```python
import numpy as np

x = np.array([1,2,0,-1,3]); k = np.array([2,-1,1]); k_T = np.array([1,-1,2])

def calculate(x, k):
    out_len = len(x) - len(k) + 1
    result = np.zeros((out_len,len(k)))
    for i in range(out_len):
        for j in range(len(k)):
            result[i,j] = x[i+j] * k[j]
    summary = np.sum(result,axis=1)
    return result,summary

M_corr, y_corr = calculate(x, k)
M_conv, y_conv = calculate(x, k_T)
print(M_corr, y_corr)
print(M_conv, y_conv)

x_test_corr = np.array([1, 2, 3, 4, 5, 6])
k_test_corr = np.array([1, -1, 2])

M_test, y_test_corr = calculate(x_test_corr, k_test_corr)

print("乘积矩阵形状：", M_test.shape)
print("输出形状：", y_test_corr.shape)
print("输出：", y_test_corr)

expected = np.array([5, 7, 9, 11])
residual = np.max(np.abs(y_test_corr - expected))

print("最大绝对残差：", residual)
print(
    "非方阵验收：",
    "PASS"
    if M_test.shape == (4, 3)
    and y_test_corr.shape == (4,)
    and residual < 1e-12
    else "FAIL"
)
```
