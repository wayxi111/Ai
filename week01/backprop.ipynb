```python
import numpy as np

x = np.array([2., 3.])          # 注意：一维数组，无批次维
W = np.array([[1., -1.],
              [0.5, 2.]])
b = np.array([0.1, -0.2])
t = np.array([1., 0.])

#计算正向输入的结果y   y = xw +b
def dense_forward(x, W, b):
    return x @ W + b

#计算反向传播的梯度dL/dx, dL/dW, dL/db   
def dense_backward(x, W, b, dy):
    y = dense_forward(x,W,b)
    dw = np.outer(x, dy)#dw = x[:, None] * dy[None, :]
    db = dy
    dx = dy @ W.T
    return dx,dw,db

def loss(x, W, b, t):
    y = dense_forward(x, W, b)
    return 0.5 * np.sum((y - t) ** 2)

def numerical_grad(f, M, eps=1e-6):
    grad = np.zeros_like(M)
    # 逐元素：M[i,j] += eps → f_plus；M[i,j] -= 2eps → f_minus；还原
    # grad[i,j] = (f_plus - f_minus) / (2 * eps)
    # 提示：f 是只吃 M 的函数（loss 要提前用偏函数绑好 x,b,t）
    for idx in np.ndindex(*M.shape):
        origin = M[idx]
        M[idx] = origin + eps
        f_plus = f(M)
        M[idx] = origin - eps
        f_minus = f(M)
        M[idx] = origin
        grad[idx] = (f_plus - f_minus) / (2 * eps)
    return grad

fW = lambda M_: loss(x, M_, b, t)
fb = lambda M_: loss(x, W, M_, t)
fx = lambda M_: loss(M_, W, b, t)

dy = dense_forward(x, W, b) - t
dx, dW, db = dense_backward(x, W, b, dy)
for name, ana, num in [('dW', dW, numerical_grad(fW, W)),
                       ('db', db, numerical_grad(fb, b)),
                       ('dx', dx, numerical_grad(fx, x))]:
    err = np.max(np.abs(ana - num))
    print(f'{name}: max err = {err:.2e}', '✓' if err < 1e-7 else '✗ 推导有错，回去查')

W1,b1 = np.array([[1.,3.],[2.,1.]]),\
        np.array([3.,2.])
W2,b2 = np.array([[5.,2.],[1.,6.]]),\
        np.array([3.,3.])

# 前向（存档！）
h1 = dense_forward(x, W1, b1)
y  = dense_forward(h1, W2, b2)

# 反向（接力）
dx2, dW2, db2 = dense_backward(h1, W2, b2, y - t)   # 层2的dx就是层1的dy
dx1, dW1, db1 = dense_backward(x, W1, b1, dx2)  # ← 边界换名！dx2 当 t 喂进来

print(h1, y)
print(dx2, dW2, db2)
print(dx1, dW1, db1)

```

```python
import numpy as np

x = np.array([2., 3.])          # 注意：一维数组，无批次维
W1, b1 = np.array([[1., 3.],
                   [2., 1.]]), np.array([3., 2.])
W2, b2 = np.array([[5., 2.],
                   [1., 6.]]), np.array([3., 3.])
t = np.array([1., 0.])

def dense_forward(x, W, b):
    return x @ W + b

# dense_backward 只管一层的责任流：给它 dy，它卸货 dW/db 并转发 dx
# dy 从哪来不是它的事：贴着 loss 的层，dy = y - t；中间层，dy = 上一层的 dx
def dense_backward(x, W, b, dy):
    dw = np.outer(x, dy)   # dw = x[:, None] * dy[None, :]
    db = dy
    dx = dy @ W.T
    return dx, dw, db

def loss(x, W, b, t):
    y = dense_forward(x, W, b)
    return 0.5 * np.sum((y - t) ** 2)

def L_total(x, W1, b1, W2, b2, t):
    first = dense_forward(x, W1, b1)
    second = dense_forward(first, W2, b2)
    return 0.5 * np.sum((second - t) ** 2)

def numerical_grad(f, M, eps=1e-4):   # eps 调大：L~6452，大 loss 的浮点噪声需要更大差分步长
    grad = np.zeros_like(M)
    for idx in np.ndindex(*M.shape):
        origin = M[idx]
        M[idx] = origin + eps
        f_plus = f(M)
        M[idx] = origin - eps
        f_minus = f(M)
        M[idx] = origin
        grad[idx] = (f_plus - f_minus) / (2 * eps)
    return grad

fW1 = lambda M_: L_total(x, M_, b1, W2, b2, t)
fb1 = lambda M_: L_total(x, W1, M_, W2, b2, t)
fx  = lambda M_: L_total(M_, W1, b1, W2, b2, t)
fW2 = lambda M_: L_total(x, W1, b1, M_, b2, t)
fb2 = lambda M_: L_total(x, W1, b1, W2, M_, t)

# 前向（存档！）
h1 = dense_forward(x, W1, b1)
y  = dense_forward(h1, W2, b2)

# 反向（接力）
dy2 = y - t                                # 只有贴着 loss 的层才有这一步（MSE 的 dL/dy）
dx2, dW2, db2 = dense_backward(h1, W2, b2, dy2)
dx1, dW1, db1 = dense_backward(x, W1, b1, dx2)  # 边界换名：层2的dx就是层1的dy

for name, ana, num in [('dW1', dW1, numerical_grad(fW1, W1)),
                       ('db1', db1, numerical_grad(fb1, b1)),
                       ('dx',  dx1, numerical_grad(fx, x)),
                       ('dW2', dW2, numerical_grad(fW2, W2)),
                       ('db2', db2, numerical_grad(fb2, b2))]:
    err = np.max(np.abs(ana - num))
    print(f'{name}: max err = {err:.2e}', '✓' if err < 1e-7 else '✗ 推导有错，回去查')
```
