```python
import numpy as np
import matplotlib.pyplot as plt

i_hat = np.array([1,0]); j_hat = np.array([0,1])

a,b = 3, 2
v = a*i_hat + b*j_hat

#固定一个标量，改变一个标量
for b_ in range(-5, 5):
    plt.plot(a*i_hat[0] + b_*j_hat[0], a*i_hat[1] + b_*j_hat[1], 'b.')
plt.axis('equal')
plt.show()

#两个标量都改变
for b_ in range(-5, 5):
    for a_ in range(-5, 5):
        plt.plot(a_*i_hat[0] + b_*j_hat[0], a_*i_hat[1] + b_*j_hat[1], 'b.')
plt.axis('equal')
plt.show()

#共线的基
u = np.array([2, 1]); w = np.array([4, 2])
for b_ in range(-5, 5):
    for a_ in range(-5, 5):
        plt.plot(a_*u[0] + b_*w[0], a_*u[1] + b_*w[1], 'b.')
plt.show()
```
